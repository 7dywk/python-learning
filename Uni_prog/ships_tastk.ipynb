```python
import numpy as np
import pandas as pd
import statistics
import matplotlib.pyplot as plt
# Probability
P = [0.13,0.17,0.15,0.25,0.2,0.1]
print(sum(P))
```

```python
# Variables

NCA = 100    # Number of Cranes Available
NSD = 175   # Number of Simulation Days
ANStL = 0  # Average Number of Ships to Load
TP = 0     # Total Probability

```

```python
# Tables

RN = np.random.rand(NSD,1) # Random Numbers https://numpy.org/doc/stable/reference/random/generated/numpy.random.rand.html
NAS = np.zeros((NSD,1))    # Number of Arriving Ships (at Port) https://numpy.org/doc/stable/reference/generated/numpy.zeros.html
NSL = np.zeros((NSD,1))    # Number of Ships to Load
NLS = np.zeros((NSD,1))    # Number of Loaded Ships
NUS = np.zeros((NSD,1))    # Number of Unloaded Ships



# Test

#print(RN[2,0])
#print(RN)
#print(NAS)

```

```python
# Day one

i = 0

if RN[0,0] >= sum(P[0:-1]):
    NAS[0,0] = 5
elif RN[0,0] >= sum(P[0:-2]):
    NAS[0,0] = 4
elif RN[0,0] >= sum(P[0:2]):
    NAS[0,0] = 3
elif RN[0,0] >= sum(P[0:1]):
    NAS[0,0] = 2
elif RN[0,0] >= sum(P[0]):
    NAS[0,0] = 1
else:
    NAS[0,0] = 0

NSL[0,0] = NAS[0,0]
NLS[0,0] = min(NSL[0,0],NCA)
NUS[0,0] = NSL[0,0] - NLS[0,0]

print(f"{i} : {round(RN[0,0],6)} : {NAS[0,0]} : {NSL[0,0]} : {NLS[0,0]} : {NUS[0,0]}")
```

```python
# Day two, three ... NSD

for i in range(1,NSD):
    if RN[i,0] >= (P[0]+P[1]+P[2]+P[3]+P[4]):
        NAS[i,0] = 5
    elif RN[i,0] >= (P[0]+P[1]+P[2]+P[3]):
        NAS[i,0] = 4
    elif RN[i,0] >= (P[0]+P[1]+P[2]):
        NAS[i,0] = 3
    elif RN[i,0] >= (P[0]+P[1]):
        NAS[i,0] = 2
    elif RN[i,0] >= P[0]:
        NAS[i,0] = 1
    else:
        NAS[i,0] = 0

    NSL[i,0] = NAS[i,0] + NUS[i-1,0]
    NLS[i,0] = min(NSL[i,0],NCA)
    NUS[i,0] = NSL[i,0] - NLS[i,0]

    print(f"{i} : {round(RN[i,0],6)} : {NAS[i,0]} : {NSL[i,0]} : {NLS[i,0]} : {NUS[i,0]}")
```

```python
ANStL = statistics.mean(NSL[:,0])
#print(f"The average number of ships to be loaded is: {ANStL})")
ANStL_rounded = round(ANStL,2)
print("The average number of ships to be loaded is: % s" % (ANStL))
print("The average number of ships to be loaded is: % s" % (ANStL_rounded))

TP = sum(P)
print('Total probability: % s' %(TP))
```
