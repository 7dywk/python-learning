# **Pi number  calculation**
## Calculating the number Pi by virtually throwing grains of sand (rice grains) onto a circle and onto a square. If a grain lands on the circle, we include it in the sum of grains in the circle - later we will substitute this number as Ac (m). All thrown grains (both those inside the circle and those outside the circle) will be counted as As (n).
## How many grains should be thrown to obtain the Pi number with the desired accuracy?


\begin{equation}
A_s = 4r^2 \Rightarrow r^2 = \frac{A_s}{4}
\end{equation}

\begin{equation}
A_c = \pi r^2 \Rightarrow r^2 = \frac{A_c}{\pi}
\end{equation}

\begin{equation}
\frac{A_s}{4} = \frac{A_c}{\pi} \Rightarrow \pi A_s = 4 A_c \Rightarrow \pi = \frac{4 A_c}{A_s}
\end{equation}

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Main code
fig = plt.figure(figsize=(6,6))
n = int(input('Enter the numer of throws [points in the simulation]: '))
m = 0 # number of points in the circle
x1 = np.random.rand(n,1)
y1 = np.random.rand(n,1)
# print(x1)
# print(y1)
x2 = x1 - 0.5
y2 = y1 - 0.5

r2 = x2**2 + y2**2
# print(r2)

for i in range(n):
    if r2[i] <= 0.25:
        m += 1
        plt.plot(x1[i], y1[i], 'r.')
    else:
        plt.plot(x1[i], y1[i], 'b.')

plt.show()

# PI calculation
My_Pi = 4*m/n

print(f'N: {n}')
print(f'Pi: {My_Pi}')
```

```python
# Main code
fig = plt.figure(figsize=(6, 6))
n = int(input('Enter the numer of throws [points in the simulation]: '))
m = 0  # number of points in the circle
x1 = np.random.rand(n, 1)
y1 = np.random.rand(n, 1)
# print(x1)
# print(y1)
x2 = x1 - 0.5
y2 = y1 - 0.5

r2 = x2 ** 2 + y2 ** 2
# print(r2)

for i in range(n):
    if r2[i] <= 0.25:
        m += 1
        plt.plot(x1[i], y1[i], 'r.')
    else:
        plt.plot(x1[i], y1[i], 'b.')

plt.show()

# PI calculation
My_Pi = 4 * m / n

print(f'N: {n}')
print(f'Pi: {My_Pi}')
import time

t_beginning = time.process_time()

t_end = time.process_time()

# Calculation of analysis time
t_analysis = t_end - t_beginning
t_analysis = round(t_analysis,1)
print(f"Analysis time: {t_analysis}s")

pi_reference_source = np.pi
pi_difference = round(My_Pi - pi_reference_source,5)
print(f"Difference: Pi - Pi referece: {pi_difference}")
```
